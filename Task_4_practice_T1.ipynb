# Задание 1
При помощи фрагмента кода определите временную
сложность операции добавления элементов в конец
списка, выполняя код для 1 000, 10 000, 100 000,
1 000 000 элементов.

```python
import time
import random

def current_milli_time():
    return round(time.time() * 1000)

# Размеры списков
sizes = [1000, 10000, 100000, 1000000]
runs = 5  # Количество запусков для усреднения

for size in sizes:
    total_time = 0  # Общее время выполнения
    
    for _ in range(runs):
        start_time = current_milli_time()
        
        # Создаём пустой список
        my_list = []
        
        # Добавляем элементы в конец списка
        for i in range(size):
            my_list.append(i)
        
        end_time = current_milli_time()
        total_time += end_time - start_time
    
    # Вычисляем среднее время выполнения
    average_time = total_time / runs
    
    # Выводим результат
    print(f"Размер списка: {size}, Среднее время выполнения: {average_time:.6f} ms")
```

# Задание 3
Напишите функцию сортировки списка методом
пузырька (Bubble Sort). Определите её сложность,
выполняя сортировку на списке разных размеров.

```python
import time
import random

def current_milli_time():
    return round(time.time() * 1000)

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]

# Размеры списков
sizes = [1000, 10000, 100000]
runs = 5  # Количество запусков для усреднения

for size in sizes:
    total_time = 0  # Общее время выполнения
    
    for _ in range(runs):
        # Создаём случайный список
        my_list = [random.randint(0, 100000) for _ in range(size)]
        
        start_time = current_milli_time()
        
        # Сортируем список методом пузырька
        bubble_sort(my_list)
        
        end_time = current_milli_time()
        total_time += end_time - start_time
    
    # Вычисляем среднее время выполнения
    average_time = total_time / runs
    
    # Выводим результат
    print(f"Размер списка: {size}, Среднее время выполнения: {average_time:.6f} ms")
```

# Задание 4
Замените функцию сортировки на стандартный
метод сортировки списка sort(). Определите
его временную сложность.

```python
import time
import random

def current_milli_time():
    return round(time.time() * 1000)

# Размеры списков
sizes = [1000, 10000, 100000, 1000000]
runs = 5  # Количество запусков для усреднения

for size in sizes:
    total_time = 0  # Общее время выполнения
    
    for _ in range(runs):
        # Создаём случайный список
        my_list = [random.randint(0, 100000) for _ in range(size)]
        
        start_time = current_milli_time()
        
        # Сортируем список стандартным методом sort()
        my_list.sort()
        
        end_time = current_milli_time()
        total_time += end_time - start_time
    
    # Вычисляем среднее время выполнения
    average_time = total_time / runs
    
    # Выводим результат
    print(f"Размер списка: {size}, Среднее время выполнения: {average_time:.6f} ms")
```

# Ответы на дополнительные вопросы

1. Что такое временная сложность алгоритма, как её определить?
Временная сложность алгоритма показывает, как растёт время выполнения алгоритма в зависимости от размера входных данных.
Её можно определить, проанализировав количество операций, выполняемых в алгоритме. Например:
O(1) — константное время (время не зависит от размера данных).
O(n) — линейное время (время растёт пропорционально размеру данных).
O(n²) — квадратичное время (время растёт пропорционально квадрату размера данных).
O(n log n) — логарифмическое время (время растёт пропорционально n × log(n)).
2. Какой способ оптимизации времени выполнения сортировки списка применён в задании №4?
В задании №4 используется стандартный метод sort(), который уже оптимизирован. Алгоритм Timsort эффективен благодаря тому, что:
- Разделяет список на небольшие отсортированные участки.
- Использует сортировку слиянием для объединения этих участков.
- Эффективно работает как с частично отсортированными, так и со случайными данными.
3. Как можно оптимизировать добавление элементов в начало списка?
Добавление элементов в начало списка имеет временную сложность O(n) , так как все элементы списка должны быть сдвинуты на одну позицию вправо.
Для оптимизации можно использовать структуры данных, такие как deque из модуля collections в Python. Добавление элемента в начало deque выполняется за O(1) , так как эта структура данных специально оптимизирована для быстрой вставки в начало и конец.

```python
import time
from collections import deque

def current_milli_time():
    """Функция для получения текущего времени в миллисекундах."""
    return round(time.time() * 1000)

# Размеры списков
sizes = [1000, 10000, 100000, 1000000]
runs = 5  # Количество запусков для усреднения

print("Сравнение времени добавления элементов в начало списка и deque:")

for size in sizes:
    # Измерение времени для обычного списка
    total_time_list = 0
    for _ in range(runs):
        my_list = []
        start_time = current_milli_time()
        
        # Добавляем элементы в начало списка
        for i in range(size):
            my_list.insert(0, i)
        
        end_time = current_milli_time()
        total_time_list += end_time - start_time
    
    average_time_list = total_time_list / runs

    # Измерение времени для deque
    total_time_deque = 0
    for _ in range(runs):
        my_deque = deque()
        start_time = current_milli_time()
        
        # Добавляем элементы в начало deque
        for i in range(size):
            my_deque.appendleft(i)
        
        end_time = current_milli_time()
        total_time_deque += end_time - start_time
    
    average_time_deque = total_time_deque / runs

    # Вывод результатов
    print(f"Размер: {size}")
    print(f"  Список (list): {average_time_list:.6f} ms")
    print(f"  Deque: {average_time_deque:.6f} ms")
```
